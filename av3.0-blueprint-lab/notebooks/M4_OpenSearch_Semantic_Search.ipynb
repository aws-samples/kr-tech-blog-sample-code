# M4: OpenSearch Semantic Search — Video Clip Retrieval

**Pipeline Position:** Stage 4: Search and Indexing — the blog's **Path A**, reduced

> **How this differs from the blog.** Stage 4 offers two paths: **Path A** = Amazon
> OpenSearch **Service**, GPU-accelerated with NVIDIA cuVS and supporting **hybrid**
> text+vector queries; **Path B** = NVIDIA **Cosmos Dataset Search** on EKS, with its own UI
> and dataset-assembly workflow. This module takes Path A but on OpenSearch
> **Serverless** (no cluster to size for a one-day workshop) and does **vector-only k-NN**,
> not hybrid. Path B is out of scope — this is not a Cosmos Dataset Search substitute.
>
> **And the vectors are a different kind.** The blog embeds each clip with **Cosmos
> Embed**, producing *joint video-text* vectors that encode visual semantics and temporal
> dynamics. This module embeds **only the caption text** with `all-MiniLM-L6-v2` (384-dim).
> So search here matches what M2's caption *says*, not what the video *shows* — cheap
> enough to run on CPU in seconds, but a real reduction in what the index can find.  
**Input S3 Path:** `s3://av30lab-user-workspace-{account_id}-{region}/users/{profile}/m2/captions.json`  
**Output S3 Path:** `s3://av30lab-user-workspace-{account_id}-{region}/users/{profile}/m4/`  
**Instance:** ml.t3.medium (CPU — no GPU required)  

> **Note:** This replaces NVIDIA Cosmos-embed NIM which requires an NVAIE (NVIDIA AI Enterprise) license.  
> OpenSearch Serverless with k-NN provides equivalent semantic search capability using AWS-native services.

## What This Module Does

1. Reads captions generated by M2 (Cosmos Reason Captioning)
2. Generates vector embeddings from caption text
3. Creates an OpenSearch Serverless k-NN vector index
4. Demonstrates natural language query → ranked video clip retrieval

## Cost Note

OpenSearch Serverless bills a **continuous 2-OCU floor** while a Classic collection exists — AWS: *"You will be billed at least for a minimum of 2 OCUs (1 OCU [0.5 x 2] indexing includes primary and standby, and 1 OCU [0.5 x 2] search includes one replica for HA) for the first collection in an account"* ([pricing](https://aws.amazon.com/opensearch-service/pricing/)). At $0.24 per OCU-hour that is **~$350/month**, not the ~$24 this note used to claim — the old figure took the 0.5 base values and dropped the ×2 for standby/replica (and did not even follow from its own premise, which gives ~$175). The cost cell at the end of this notebook has always computed the correct ~$350.

**This notebook does not delete the collection it creates** — it has no `delete_collection` call, so the 2-OCU floor keeps running after the last cell finishes. Cleanup is somebody else's job: `delete_user` tears it down when the participant is removed, and `scripts/teardown.sh` sweeps for orphans. A forgotten collection is roughly **$350/month**, not pocket change, so if you are running this module outside a managed workshop profile, delete it yourself — the cost cell at the end of this notebook prints the exact `aws opensearchserverless delete-collection` command. (AWS also offers a dev-test deployment without standby replicas, and NextGen collections that scale to zero when idle — either would cut this floor, but neither is what this module creates.)

```python
"""Load M2 captions — these are the documents we will index"""
from botocore.exceptions import ClientError

captions_key = f"{INPUT_PREFIX}captions.json"
print(f"Loading captions from: s3://{USER_BUCKET}/{captions_key}")

try:
    response = s3.get_object(Bucket=USER_BUCKET, Key=captions_key)
except ClientError as e:
    code = e.response.get("Error", {}).get("Code", "")
    if code in ("NoSuchKey", "404", "NoSuchBucket", "AccessDenied"):
        raise RuntimeError(
            f"M2 captions not found at s3://{USER_BUCKET}/{captions_key} "
            f"({code}). Run M2 (Cosmos Reason Captioning) first — it writes "
            f"captions.json that M4 indexes."
        ) from e
    raise

m2_output = json.loads(response["Body"].read())
captions = m2_output.get("captions")
if not captions:
    raise RuntimeError(
        f"M2 output at s3://{USER_BUCKET}/{captions_key} has no 'captions' — "
        f"re-run M2 to regenerate it."
    )
print(f"Loaded {len(captions)} captions from M2")
print(f"Model used: {m2_output.get('model', 'unknown')}")
print(f"Generated at: {m2_output.get('generated_at', 'unknown')}")

print(f"\nSample documents:")
for i, cap in enumerate(captions[:3]):
    print(f"  [{i}] {cap.get('filename', '?')}: {cap.get('caption', '')[:80]}...")
```

```python
"""Generate embeddings from captions using sentence-transformers"""

MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"  # 384-dim, fast on CPU


class MiniLMEncoder:
    """The slice of the sentence-transformers API this notebook uses, on plain transformers.

    Mean-pool the token embeddings under the attention mask, then L2-normalise — which is
    exactly what sentence-transformers does for THIS model: its 1_Pooling/config.json sets
    pooling_mode_mean_tokens true and every other mode false, word_embedding_dimension 384,
    and sentence_bert_config.json sets max_seq_length 256. See the install cell above for why
    the library itself is not installed.
    """

    def __init__(self, model_name: str = MODEL_NAME):
        self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        self.model = AutoModel.from_pretrained(model_name).eval()

    @torch.no_grad()
    def encode(self, sentences, batch_size: int = 32, show_progress_bar: bool = False,
               normalize_embeddings: bool = False):
        if isinstance(sentences, str):
            sentences = [sentences]
        chunks = []
        for start in range(0, len(sentences), batch_size):
            batch = sentences[start:start + batch_size]
            enc = self.tokenizer(batch, padding=True, truncation=True,
                                 max_length=256, return_tensors="pt")
            hidden = self.model(**enc).last_hidden_state           # (B, T, 384)
            # Mask BEFORE summing: padding tokens carry real activations and would otherwise
            # pull every short caption's vector toward the padding direction.
            mask = enc["attention_mask"].unsqueeze(-1).to(hidden.dtype)
            pooled = (hidden * mask).sum(dim=1) / mask.sum(dim=1).clamp(min=1e-9)
            if normalize_embeddings:
                pooled = torch.nn.functional.normalize(pooled, p=2, dim=1)
            chunks.append(pooled)
            if show_progress_bar:
                done = min(start + batch_size, len(sentences))
                print(f"    encoded {done}/{len(sentences)}")
        return torch.cat(chunks).numpy()


print(f"Loading embedding model: {MODEL_NAME}")
start_embed = time.time()
try:
    embed_model = MiniLMEncoder()
except Exception as e:
    raise RuntimeError(
        f"Failed to load the embedding model '{MODEL_NAME}': {e}\n"
        f"This model is public (no HF token needed) but must be downloaded once — "
        f"the instance needs outbound network egress to huggingface.co on the first "
        f"run. Check the workspace has internet egress, then re-run."
    ) from e
print(f"Model loaded in {time.time() - start_embed:.1f}s")

# Generate embeddings for all captions
texts = [cap["caption"] for cap in captions]
print(f"\nGenerating embeddings for {len(texts)} captions...")

start_encode = time.time()
embeddings = embed_model.encode(texts, show_progress_bar=True, normalize_embeddings=True)
encode_time = time.time() - start_encode

print(f"\nEmbedding generation complete:")
print(f"  Shape: {embeddings.shape}")
print(f"  Dimension: {embeddings.shape[1]}")
print(f"  Time: {encode_time:.2f}s")
print(f"  Per caption: {encode_time/len(texts)*1000:.1f}ms")

# Verify normalization
norms = np.linalg.norm(embeddings, axis=1)
print(f"  L2 norm range: [{norms.min():.4f}, {norms.max():.4f}] (should be ~1.0)")
```

```python
"""Create OpenSearch Serverless collection (or connect to pre-provisioned one)"""

def _iam_principal_arn():
    """AOSS data-access policies match on the IAM role/user ARN, NOT the STS
    assumed-role SESSION ARN. get_caller_identity returns the session form
    (arn:aws:sts::<acct>:assumed-role/<RoleName>/<session>); convert it to the
    IAM role ARN (arn:aws:iam::<acct>:role/<RoleName>) so index writes/searches
    under the SageMaker execution role are authorized instead of 403."""
    ident = boto3.client("sts").get_caller_identity()
    arn = ident["Arn"]
    acct = ident["Account"]
    if ":assumed-role/" in arn:
        role_name = arn.split(":assumed-role/", 1)[1].split("/", 1)[0]
        return f"arn:aws:iam::{acct}:role/{role_name}"
    return arn  # already an iam user/role ARN


def get_or_create_collection(collection_name: str) -> dict:
    """Get existing collection or create a new VECTORSEARCH collection."""

    # Check if collection already exists. Narrow the except so a real
    # permission error surfaces instead of silently falling through to create.
    try:
        response = aoss_client.batch_get_collection(names=[collection_name])
        if response.get("collectionDetails"):
            collection = response["collectionDetails"][0]
            print(f"Found existing collection: {collection['name']}")
            print(f"  Status: {collection['status']}")
            print(f"  Endpoint: {collection.get('collectionEndpoint', 'N/A')}")
            return collection
    except aoss_client.exceptions.ResourceNotFoundException:
        pass
    except Exception as e:
        print(f"Collection lookup (continuing to create): {e}")

    # Create encryption policy (required before collection)
    encryption_policy = json.dumps({
        "Rules": [{"ResourceType": "collection", "Resource": [f"collection/{collection_name}"]}],
        "AWSOwnedKey": True
    })

    try:
        aoss_client.create_security_policy(
            name=f"{collection_name}-enc",
            type="encryption",
            policy=encryption_policy
        )
        print("Created encryption policy.")
    except aoss_client.exceptions.ConflictException:
        print("Encryption policy already exists.")

    # Create network policy (public access for workshop)
    # Collection-only network rule (no public "dashboard"/web-UI rule).
    # Data access is still gated by IAM (aoss:APIAccessAll) + the data-
    # access policy naming this execution role, so the endpoint being
    # public does not permit unauthenticated reads/writes.
    network_policy = json.dumps([{
        "Rules": [{
            "ResourceType": "collection",
            "Resource": [f"collection/{collection_name}"]
        }],
        "AllowFromPublic": True
    }])

    try:
        aoss_client.create_security_policy(
            name=f"{collection_name}-net",
            type="network",
            policy=network_policy
        )
        print("Created network policy.")
    except aoss_client.exceptions.ConflictException:
        print("Network policy already exists.")

    # Create data access policy. Principal MUST be the IAM role ARN (not the STS
    # session ARN) or index writes/searches 403 under the execution role.
    caller_arn = _iam_principal_arn()
    print(f"Data-access principal: {caller_arn}")
    data_policy = json.dumps([{
        "Rules": [{
            "ResourceType": "index",
            "Resource": [f"index/{collection_name}/*"],
            # aoss:DeleteIndex is load-bearing, not completeness: without it the index
            # cell's delete fails with 403 authorization_exception, and because AOSS
            # rejects a caller-supplied document _id every index() call INSERTS — so
            # each re-run of this module adds another full copy of the captions.
            # Measured on 2026-09-28 across five runs of one participant's collection:
            # 24 -> 48 -> 72 -> 96 -> 120 documents for 24 captions.
            "Permission": ["aoss:CreateIndex", "aoss:UpdateIndex", "aoss:DescribeIndex",
                           "aoss:DeleteIndex",
                           "aoss:ReadDocument", "aoss:WriteDocument"]
        }, {
            "ResourceType": "collection",
            "Resource": [f"collection/{collection_name}"],
            "Permission": ["aoss:CreateCollectionItems", "aoss:DescribeCollectionItems",
                           "aoss:UpdateCollectionItems", "aoss:DeleteCollectionItems"]
        }],
        "Principal": [caller_arn]
    }])

    try:
        aoss_client.create_access_policy(
            name=f"{collection_name}-access",
            type="data",
            policy=data_policy
        )
        print("Created data access policy.")
    except aoss_client.exceptions.ConflictException:
        print("Data access policy already exists.")
        # An existing policy may predate the aoss:DeleteIndex grant above, in which
        # case the index cell cannot clear the index and this module duplicates its
        # documents on every run. This role can READ the policy (aoss:GetAccessPolicy)
        # but not update it: aoss:UpdateAccessPolicy is deliberately NOT granted,
        # because on collection/* it would let one participant grant themselves access
        # to another participant's collection. So check and report precisely rather
        # than leaving the operator to guess.
        try:
            _pol = aoss_client.get_access_policy(
                name=f"{collection_name}-access", type="data"
            )["accessPolicyDetail"]["policy"]
            if isinstance(_pol, str):
                _pol = json.loads(_pol)
            _perms = {
                p
                for stmt in (_pol if isinstance(_pol, list) else [_pol])
                for rule in stmt.get("Rules", [])
                for p in rule.get("Permission", [])
            }
            if "aoss:DeleteIndex" not in _perms:
                print("  NOTE: this policy predates the aoss:DeleteIndex grant, so the")
                print("        index cell below CANNOT clear the index — its delete gets")
                print("        403 — and every re-run of this module ADDS another copy of")
                print("        all documents, which makes the search demo rank duplicates.")
                print("        This role cannot update the policy (by design). For a clean")
                print("        index an admin must delete BOTH the collection and the")
                print(f"        '{collection_name}-access' data policy; this cell then")
                print("        recreates them with the delete grant. Commands are in")
                print("        docs/en/ADMIN_GUIDE.md (search: DeleteIndex).")
        except Exception as _e:
            print(f"  (could not read the existing policy to check it: {_e})")

    # Create the collection
    print(f"\nCreating collection: {collection_name} (type=VECTORSEARCH)")
    try:
        response = aoss_client.create_collection(
            name=collection_name,
            type="VECTORSEARCH",
            description="AV 3.0 Blueprint Lab - Semantic video clip search"
        )
    except Exception as e:
        raise RuntimeError(
            f"Failed to create AOSS collection '{collection_name}': {e}\n"
            f"Common causes: the account AOSS collection quota is reached, or the "
            f"execution role lacks aoss:CreateCollection. Check Service Quotas "
            f"(OpenSearch Serverless) and the role policy."
        ) from e

    collection = response["createCollectionDetail"]
    print(f"  Collection ID: {collection['id']}")
    print(f"  Status: {collection['status']}")

    # Wait for collection to become ACTIVE
    print("\nWaiting for collection to become ACTIVE (this may take 2-5 minutes)...")
    for attempt in range(30):
        time.sleep(10)
        status_response = aoss_client.batch_get_collection(ids=[collection["id"]])
        current = status_response["collectionDetails"][0]
        status = current["status"]
        print(f"  [{attempt*10}s] Status: {status}")
        if status == "ACTIVE":
            print(f"\nCollection is ACTIVE!")
            print(f"  Endpoint: {current['collectionEndpoint']}")
            return current
        elif status == "FAILED":
            raise RuntimeError(f"Collection creation failed: {current}")

    raise TimeoutError("Collection did not become ACTIVE within 5 minutes.")

collection = get_or_create_collection(COLLECTION_NAME)
ENDPOINT = collection["collectionEndpoint"]

# AOSS data-access-policy changes are NOT effective the moment the collection is
# ACTIVE, so a request issued immediately can still 403. AWS documents the figure:
# "There might be about a minute of lag time between when you create the policy and
# when the permissions are enforced. If it takes more than 5 minutes, contact Support."
# (serverless-data-access.html) — and "a few minutes" for a policy UPDATE. An earlier
# version of this comment said "a few seconds", which understated it by an order of
# magnitude. This sleep is only a head start; what actually covers the documented
# minute is the retry loop in the index cell below.
time.sleep(15)
```

```python
"""Connect to OpenSearch Serverless and create k-NN vector index"""

# Build authenticated OpenSearch client
region = boto3.session.Session().region_name or "us-west-2"
credentials = boto3.Session().get_credentials()
if credentials is None:
    raise RuntimeError(
        "No AWS credentials available in this kernel — cannot sign OpenSearch "
        "requests. In SageMaker Studio the execution role is normally present; "
        "check the space's role configuration."
    )
# Sign with the credentials OBJECT, not a snapshot of its strings.
#
# This used to be AWS4Auth(credentials.access_key, credentials.secret_key, ...).
# Reading .access_key off a SageMaker execution role FREEZES the value: the role's
# credentials are temporary and botocore rotates them behind RefreshableCredentials,
# but the frozen strings keep being used to sign. Once they expire, every request is
# signed with dead keys and AOSS answers a bare 403 Forbidden — with a data-access
# policy that is completely correct, which is exactly how much time that costs to
# diagnose. Measured 2026-09-28: writes in one cell succeeded, then every read in a
# later cell returned 403 — search, count, get_mapping and cat.indices alike — while
# GetAccessPolicy showed aoss:ReadDocument granted to this very role.
#
# AWSV4SignerAuth holds the credentials object and re-resolves per request, so a
# rotation is invisible. This is the client AWS documents for serverless collections.
if AWSV4SignerAuth is not None:
    awsauth = AWSV4SignerAuth(credentials, region, "aoss")
else:
    _frozen = credentials.get_frozen_credentials()
    awsauth = AWS4Auth(
        _frozen.access_key,
        _frozen.secret_key,
        region,
        "aoss",
        session_token=_frozen.token,
    )
    print("WARNING: this opensearch-py has no AWSV4SignerAuth, so the credentials are")
    print("         frozen at this moment. If a later cell starts returning 403 on")
    print("         reads or writes, re-run THIS cell to re-sign with fresh ones.")

# Strip https:// prefix for host
host = ENDPOINT.replace("https://", "")

os_client = OpenSearch(
    hosts=[{"host": host, "port": 443}],
    http_auth=awsauth,
    use_ssl=True,
    verify_certs=True,
    connection_class=RequestsHttpConnection,
    timeout=60
)

# Create k-NN index
index_body = {
    "settings": {
        "index": {
            "knn": True,
            "knn.algo_param.ef_search": 512
        }
    },
    "mappings": {
        "properties": {
            "caption_embedding": {
                "type": "knn_vector",
                "dimension": EMBEDDING_DIM,
                "method": {
                    "name": "hnsw",
                    "space_type": "cosinesimil",
                    "engine": "faiss",
                    "parameters": {
                        "ef_construction": 512,
                        "m": 16
                    }
                }
            },
            "caption": {"type": "text"},
            "filename": {"type": "keyword"},
            "frame_idx": {"type": "integer"},
            "timestamp": {"type": "date"},
            # Stamped with one id per run so the indexing cell can count exactly the
            # documents IT wrote. keyword, not text: a term query on an analysed text
            # field would not match the id. See the indexing cell for why counting the
            # index total does not work.
            "run_id": {"type": "keyword"}
        }
    }
}

# Drop any previous copy of the index first, so a re-run does not accumulate
# duplicates. This is not tidiness: AOSS refuses a caller-supplied document _id
# ("Document ID is not supported in create/index operation request", see the
# indexing cell), so every index() call INSERTS. Re-running the module therefore
# added another copy of all 24 captions each time — measured across four runs on
# 2026-09-28: 24 -> 48 -> 72 -> 96 documents for 24 captions, and the k-NN demo then
# ranked the same caption several times. The documents are rebuilt from M2's captions
# on every run, so deleting the index loses nothing.
#
# The delete is VERIFIED, not announced. A first attempt did `delete` + `sleep(5)` +
# print "Deleted" and the documents survived anyway, because OpenSearch Serverless is
# eventually consistent on the read path: this collection took 61 s for 24 freshly
# indexed documents to become countable, so 5 s says nothing about whether a delete
# landed. Poll until the index is gone (count raises index_not_found) or reports 0.
DELETE_TIMEOUT_S = 150
_del_err = None
try:
    os_client.indices.delete(index=INDEX_NAME)
except Exception as _e:
    if "index_not_found" in str(_e) or "404" in str(_e):
        _del_err = "absent"          # nothing to delete — the normal first run
    else:
        _del_err = str(_e)

if _del_err == "absent":
    print(f"No previous index to delete: {INDEX_NAME}")
elif _del_err:
    print(f"WARNING: could not delete the existing index: {_del_err}")
    print(f"         Documents will be DUPLICATED on top of whatever is already there.")
else:
    _t0 = time.time()
    _cleared = False
    while time.time() - _t0 < DELETE_TIMEOUT_S:
        try:
            _n = os_client.count(index=INDEX_NAME)["count"]
        except Exception as _e:
            if "index_not_found" in str(_e) or "404" in str(_e):
                _cleared = True      # the index is really gone
                break
            _n = -1
        if _n == 0:
            _cleared = True
            break
        time.sleep(5)
    if _cleared:
        print(f"Deleted the previous index, VERIFIED empty after "
              f"{time.time() - _t0:.0f}s: {INDEX_NAME}")
    else:
        print(f"WARNING: the delete of {INDEX_NAME} was accepted but the index still")
        print(f"         reports documents after {DELETE_TIMEOUT_S}s. The indexing cell")
        print(f"         will report how many it found, and they WILL be duplicated.")
        print(f"         Cleanest fix: delete the whole collection and re-run this module")
        print(f"         (the cost cell prints the delete-collection command).")

# Create the index. aoss does not support indices.exists (404), so we just try
# to create it and treat "already exists" as fine.
#
# RETRY: a freshly-created collection's data-access policy takes time to
# propagate to the data plane — an index create issued too soon returns
# 403 "User does not have permissions for the requested resource" even though
# the policy names this exact role. That propagation can take well over the
# collection-cell's initial wait, so we retry the create with backoff and treat
# 403 as "policy still propagating" rather than a hard failure.
_created = False
_last_err = None
for _attempt in range(12):   # ~ up to ~2.5 min total (10s..~25s backoff, capped)
    try:
        os_client.indices.create(index=INDEX_NAME, body=index_body)
        _created = True
        break
    except Exception as _e:
        _msg = str(_e)
        if "resource_already_exists" in _msg or "already exists" in _msg:
            print(f"Index already exists: {INDEX_NAME}")
            _created = True
            break
        # 403 here == the data-access policy has not propagated yet; wait + retry.
        if "403" in _msg or "authorization_exception" in _msg or "does not have permissions" in _msg:
            _last_err = _e
            _wait = min(10 + _attempt * 3, 25)
            print(f"  [{_attempt+1}/12] AOSS policy still propagating (403) — "
                  f"retrying in {_wait}s...")
            time.sleep(_wait)
            continue
        raise   # any other error is a real failure — surface it
if not _created:
    raise RuntimeError(
        f"Could not create index '{INDEX_NAME}' after retries — the AOSS "
        f"data-access policy for this collection never propagated (last error "
        f"below). The policy principal is correct (see the collection cell); this "
        f"is a propagation delay. Wait a minute and re-run THIS cell.\n"
        f"  Last error: {_last_err}"
    )
print(f"Created k-NN index: {INDEX_NAME}")
print(f"  Dimension: {EMBEDDING_DIM}")
print(f"  Space type: cosinesimil")
print(f"  Engine: faiss (HNSW)")
```

```python
"""Index all caption embeddings into OpenSearch"""

print(f"Indexing {len(captions)} documents...")

import uuid

RUN_ID = uuid.uuid4().hex          # stamped on every document this run writes


def _visible_docs(run_id=None):
    """Documents visible to SEARCH — all of them, or only this run's.

    WHY run_id EXISTS. The wait at the end of this cell has to answer "are the documents
    I just wrote searchable yet", and counting cannot answer it, in either direction:

      - Comparing the index TOTAL against this run's count is satisfied vacuously the
        moment the index is not empty. Measured 2026-09-28: a re-run printed
        "Searchable: 48/24 after 0s" — 48 >= 24 was already true, so it never waited.
      - Adding a "documents before" reading does not fix it, because that reading is
        itself unreliable. AOSS's read path lags a delete: the index cell deletes and
        recreates the index, and a probe issued straight afterwards can still report the
        PRE-DELETE count. Measured 2026-09-28: docs_before read 24 for an index that had
        just been recreated empty, the target became 24 + 24 = 48, and the wait sat at
        24/48 for its full 181s because 48 documents never existed.

    Counting this run's own stamp is exact and immune to both: leftovers do not inflate
    it, and a stale pre-delete view does not either, because those documents carry a
    different run_id.

    The probe uses _search rather than _count because gating on the operation you are
    about to perform is the right shape: if the demo below can run, the probe can run.
    (An earlier comment here claimed AOSS refused _count while allowing _search. That was
    unsound — the failing _count had been signed with expired credentials, which the
    client cell above now fixes with a refreshing signer.)

    request_timeout caps ONE call (the client is built with timeout=60). The fallback
    covers an opensearch-py that does not accept the kwarg — letting that TypeError reach
    the caller's except would read as "0 documents visible" forever.
    """
    _q = {"term": {"run_id": run_id}} if run_id else {"match_all": {}}
    _body = {"size": 0, "query": _q, "track_total_hits": True}
    try:
        _res = os_client.search(index=INDEX_NAME, body=_body, request_timeout=10)
    except TypeError:
        _res = os_client.search(index=INDEX_NAME, body=_body)
    _total = _res["hits"]["total"]
    return _total["value"] if isinstance(_total, dict) else int(_total)


# Informational only — it does NOT feed the wait below any more. Its value can be a
# stale pre-delete reading, which is exactly how the 24/48 stall above happened.
docs_before = None
try:
    docs_before = _visible_docs()
except Exception as _e:
    _msg = f"{type(_e).__name__}: {_e}"
    if "index_not_found" in _msg or "404" in _msg:
        # Expected, and it says what we want to know: the index cell just recreated the
        # index and a brand-new index is not on the search path yet — the same readiness
        # lag that makes documents take ~60s to appear. "No such index" positively states
        # that nothing is searchable.
        print(f"  Index is not on the search path yet (the cell above just recreated "
              f"it) — nothing already searchable, as expected.")
        docs_before = 0
    else:
        print(f"  (could not read the existing document count: {_msg[:110]} — "
              f"continuing; the wait below does not depend on it)")
if docs_before:
    print(f"  NOTE: the index reports {docs_before} document(s) already searchable.")
    print(f"  That is EITHER genuinely leftover documents — in which case the search demo")
    print(f"  may rank some captions twice — OR a stale read of the index the cell above")
    print(f"  just deleted, because AOSS's read path lags a delete. Check that cell's")
    print(f"  output: if it said VERIFIED empty, this is the stale reading and harmless.")
    print(f"  Either way the wait below is unaffected: it counts only this run's own")
    print(f"  documents, by run_id.")
start_index = time.time()

indexed_count = 0
errors = []          # each entry keeps the DOCUMENT so the retry pass below can resend it

for i, (cap, embedding) in enumerate(zip(captions, embeddings)):
    doc = {
        "caption_embedding": embedding.tolist(),
        "caption": cap["caption"],
        "filename": cap["filename"],
        "frame_idx": cap.get("frame_idx", i),
        "timestamp": cap.get("timestamp", datetime.now(timezone.utc).isoformat()),
        "run_id": RUN_ID,
    }
    try:
        # AOSS (OpenSearch Serverless) does NOT support a caller-supplied
        # document _id on index/create — it auto-generates one. Passing id=
        # raises 400 illegal_argument_exception ("Document ID is not supported
        # in create/index operation request"). Omit it; AOSS assigns the _id.
        os_client.index(
            index=INDEX_NAME,
            body=doc,
        )
        indexed_count += 1
    except Exception as e:
        errors.append({"idx": i, "filename": cap.get("filename"), "doc": doc,
                       "error": f"{type(e).__name__}: {e}"})
first_pass_indexed = indexed_count

index_time = time.time() - start_index
print(f"\nIndexing complete:")
print(f"  Documents indexed: {indexed_count}/{len(captions)}")
print(f"  Errors: {len(errors)}")
print(f"  Time: {index_time:.2f}s")
if first_pass_indexed:
    print(f"  Throughput: {first_pass_indexed/index_time:.1f} docs/s")

# A PARTIAL failure used to pass in silence. The only guard was `indexed_count == 0`,
# so a run that indexed 23 of 24 printed "Errors: 1", kept the error text to itself,
# and went on to report the searchable wait as a success — measured 2026-09-28. One
# missing caption is invisible in a k-NN demo (it just never ranks) and would have
# shipped that way in the published bundle. So: show every failure, retry each once —
# a 429 or a cold-collection hiccup is exactly what succeeds on a second attempt —
# and only then decide.
if errors:
    print(f"\n  {len(errors)} document(s) failed on the first pass:")
    for _f in errors[:5]:
        print(f"    [{_f['idx']}] {_f.get('filename')}: {_f['error'][:150]}")
    if len(errors) > 5:
        print(f"    ... and {len(errors) - 5} more")
    # Name the cause when every failure is a 5xx, so nine red lines do not send the
    # next operator investigating. Measured across runs on a freshly recreated index:
    # 1 failure at idx 0, then 9 at idx 0,2,4,6,8,... — every other document. A server
    # -side 500 in an alternating pattern is the index still coming up, not a problem
    # with the documents, and it is why the retry happens AFTER the whole pass rather
    # than inline: by then the index is warm. Both runs recovered fully on retry.
    if all("(50" in _f["error"] or "TransportError" in _f["error"] for _f in errors):
        print(f"  All {len(errors)} are server-side 5xx on an index this cell's"
              f" predecessor just recreated —")
        print(f"  the index is still coming up. Expected, and what the retry below is"
              f" for. Nothing is")
        print(f"  wrong with the documents; if the retry clears them the run is clean.")
    print(f"  Retrying them once...")
    time.sleep(5)
    _still = []
    for _f in errors:
        try:
            os_client.index(index=INDEX_NAME, body=_f["doc"])
            indexed_count += 1
        except Exception as _e:
            _f["retry_error"] = f"{type(_e).__name__}: {_e}"
            _still.append(_f)
    errors = _still
    print(f"  After retry: {indexed_count}/{len(captions)} indexed, "
          f"{len(errors)} still failing")

# Anything short of every caption is fatal. This module's contract is to index all of
# M2's captions; a partial index quietly narrows the search demo, which is worse than
# a loud stop because nothing downstream can tell.
if indexed_count < len(captions):
    _detail = "\n".join(
        f"  [{_f['idx']}] {_f.get('filename')}: {_f.get('retry_error', _f['error'])}"
        for _f in errors[:5]) or "  (no error captured)"
    raise RuntimeError(
        f"Indexed {indexed_count}/{len(captions)} documents — {len(errors)} still "
        f"failing after one retry, so the search demo would run over an incomplete "
        f"index.\n{_detail}\n"
        f"If these are 403/auth_error: the AOSS data-access policy principal may not "
        f"match this execution role (see the collection cell), or the policy is still "
        f"propagating — AWS documents about a minute after creation.\n"
        f"If they are 429/throttling or connection errors: re-run from the INDEX cell "
        f"above, not this one. That cell deletes and recreates the index, so a re-run "
        f"starts clean instead of adding a second copy of the documents that did land."
    )

# Wait until the documents are actually SEARCHABLE, not merely accepted.
#
# OpenSearch Serverless has no _refresh API (indices.refresh -> 404); it makes new
# documents searchable on its own schedule. This used to be a flat `time.sleep(10)`,
# and 10s was not enough: the captured run in examples/notebooks-with-outputs.tar.gz
# indexed 24/24 and then every one of the five demo queries in the next cell returned
# ZERO hits, because the search ran before the index was searchable. _count reflects
# the searchable view, so poll it instead of guessing a delay.
SEARCHABLE_TIMEOUT_S = 180
want_visible = indexed_count        # this run's documents only — see _visible_docs

# Report progress while waiting, and never wait out the clock on an error that is not
# going to fix itself. The first version printed nothing until the loop finished, so a
# normal wait was indistinguishable from a hang; and because it swallowed every
# exception, a 403 on the probe spun the full 180s in silence. Both are fixed: the
# count is printed on the first poll, when it CHANGES, and every ~30s; a failing probe
# prints its error; and three identical failures in a row stop the wait instead of
# burning the remaining timeout.
# AOSS has no _refresh API, and for vector search collections AWS documents the
# reason this takes so long: "The refresh interval for indexes on vector search
# collections is 60 seconds." (serverless-vector-search.html; NextGen vector
# collections are 10s). Measured here: 20-61s, which matches. So a wait of tens of
# seconds is the designed behaviour, not a symptom.
print(f"  Waiting for {want_visible} document(s) to become searchable — vector search"
      f" collections refresh on a documented 60s interval (measured 20-61s here)...")
_wait_start = time.time()
visible = 0
_poll = 0
_last_reported = None
_fail_streak = 0
_last_err = ""
while time.time() - _wait_start < SEARCHABLE_TIMEOUT_S:
    _poll += 1
    _failed = False
    try:
        visible = _visible_docs(RUN_ID)
        _note = ""
        _fail_streak = 0
    except Exception as _e:
        visible = 0
        _err = f"{type(_e).__name__}: {str(_e)[:60]}"
        _note = f"   ({_err})"
        _fail_streak = _fail_streak + 1 if _err == _last_err else 1
        _last_err = _err
        _failed = True
    if _failed or _poll == 1 or visible != _last_reported or _poll % 6 == 0:
        print(f"    t+{time.time() - _wait_start:>3.0f}s  count={visible}/{want_visible}{_note}")
        _last_reported = visible
    if visible >= want_visible:
        break
    if _fail_streak >= 3:
        _left = SEARCHABLE_TIMEOUT_S - (time.time() - _wait_start)
        print(f"    The probe failed {_fail_streak}x with the same error. Not waiting out the")
        print(f"    remaining {_left:.0f}s — this will not fix itself.")
        print(f"    The documents ARE indexed ({indexed_count}/{len(captions)}, 0 errors);")
        print(f"    it is the READ path that is refusing. Two known causes, in order of")
        print(f"    likelihood: (1) the data-access policy has not finished propagating —")
        print(f"    it is minutes old if this collection was just created, and the index")
        print(f"    cell above retries the same 403 for up to 2.5 min for this reason, so")
        print(f"    simply re-running this cell often clears it; (2) the policy genuinely")
        print(f"    does not grant aoss:ReadDocument to this role — check it with")
        print(f"    get_access_policy. Expired credentials USED to be the usual answer")
        print(f"    here; the client cell now signs with a refreshing signer, so that one")
        print(f"    is off the table.")
        print(f"    To tell the two apart, look for the 'x-aoss-response-hint' response")
        print(f"    header: AWS documents 'X01:gw-helper-deny' there for the case where the")
        print(f"    principal is missing the IAM-side permission, as opposed to a policy")
        print(f"    that is still propagating (repost.aws/knowledge-center/")
        print(f"    opensearch-serverless-troubleshoot).")
        break
    time.sleep(5)
print(f"  Searchable: {visible}/{want_visible} after {time.time() - _wait_start:.0f}s")
if visible < want_visible:
    print(f"  WARNING: only {visible} of {want_visible} documents are searchable after "
          f"{time.time() - _wait_start:.0f}s — the demo below may return few or no hits. The next "
          f"cell fails loudly rather than printing an empty success, so just re-run it.")
```

```python
"""Demo: Natural language queries → ranked video clip results"""

def semantic_search(query: str, k: int = 5) -> list:
    """Search for video clips using natural language."""
    # Encode query
    query_embedding = embed_model.encode([query], normalize_embeddings=True)[0]

    # k-NN search
    search_body = {
        "size": k,
        "query": {
            "knn": {
                "caption_embedding": {
                    "vector": query_embedding.tolist(),
                    "k": k
                }
            }
        },
        "_source": ["caption", "filename", "frame_idx"]
    }

    try:
        response = os_client.search(index=INDEX_NAME, body=search_body)
    except Exception as e:
        raise RuntimeError(
            f"OpenSearch query failed: {e}\n"
            f"If this is a 403/auth_error, the data-access policy principal may not "
            f"match this role; if it is 'index_not_found', indexing did not complete."
        ) from e

    results = []
    for hit in response["hits"]["hits"]:
        results.append({
            "score": hit["_score"],
            "filename": hit["_source"]["filename"],
            "frame_idx": hit["_source"].get("frame_idx"),
            "caption": hit["_source"]["caption"]
        })

    return results


# Demo queries representing AV use cases
DEMO_QUERIES = [
    "pedestrian crossing the road at an intersection",
    "highway driving with multiple lanes and vehicles ahead",
    "rainy weather conditions with reduced visibility",
    "parked cars on both sides of a narrow street",
    "traffic signal turning red with approaching vehicles"
]

print("=" * 70)
print("Semantic Video Clip Search — Demo Results")
print("=" * 70)

all_results = {}
for query in DEMO_QUERIES:
    print(f"\nQuery: \"{query}\"")
    print("-" * 50)

    results = semantic_search(query, k=3)
    all_results[query] = results

    for rank, r in enumerate(results, 1):
        print(f"  #{rank} [score={r['score']:.4f}] {r['filename']}")
        print(f"      {r['caption'][:100]}...")

print("\n" + "=" * 70)

# Fail loudly on an empty demo. The indexing cell already guards "0 documents
# indexed" precisely so the search does not silently return nothing; the same
# reasoning has to apply here. Without this guard the captured run in
# examples/notebooks-with-outputs.tar.gz printed "Search demo complete" underneath
# five EMPTY result sets, and the next cell wrote them to S3 as this module's
# deliverable and validated them as OK.
total_hits = sum(len(r) for r in all_results.values())
if total_hits == 0:
    raise RuntimeError(
        f"All {len(DEMO_QUERIES)} demo queries returned 0 hits, so this module "
        f"demonstrated nothing.\n"
        f"Most likely the index was not searchable yet — OpenSearch Serverless has no "
        f"_refresh API and makes documents searchable on its own schedule. The previous "
        f"cell now waits for that and prints 'Searchable: N/N'; if it printed a WARNING, "
        f"just RE-RUN THIS CELL — the documents are already indexed, so it is free.\n"
        f"If it stays empty: confirm the previous cell reported 'Documents indexed: N/N', "
        f"then check the index name ({INDEX_NAME!r}) and that the k-NN field "
        f"'caption_embedding' matches the mapping created two cells above."
    )

print(f"Search demo complete — {total_hits} hits across {len(DEMO_QUERIES)} queries, "
      f"all via k-NN vector similarity.")
```

```python
"""Cost Analysis — ml.t3.medium + OpenSearch Serverless"""

INSTANCE_TYPE = "ml.t3.medium"
INSTANCE_COST_PER_HOUR = 0.05  # USD, us-west-2 on-demand (adjust for your region)
KRW_RATE = 1370

# OpenSearch Serverless pricing
AOSS_OCU_COST_PER_HOUR = 0.24  # USD per OCU-hour
# AOSS bills a continuous per-account OCU floor 24/7. With standby
# replicas (the create default) the practical minimum is ~1 OCU for
# indexing + ~1 OCU for search = ~2 OCU while the collection exists.
AOSS_MIN_INDEXING_OCU = 1.0
AOSS_MIN_SEARCH_OCU = 1.0
AOSS_MONTHLY_MIN = (AOSS_MIN_INDEXING_OCU + AOSS_MIN_SEARCH_OCU) * AOSS_OCU_COST_PER_HOUR * 730  # hours/month

# Execution time estimate
estimated_total_min = 15  # encoding + indexing + queries
estimated_hours = estimated_total_min / 60

compute_cost_usd = INSTANCE_COST_PER_HOUR * estimated_hours
aoss_cost_per_hour = (AOSS_MIN_INDEXING_OCU + AOSS_MIN_SEARCH_OCU) * AOSS_OCU_COST_PER_HOUR
aoss_session_cost = aoss_cost_per_hour * estimated_hours  # only while active

total_session_cost_usd = compute_cost_usd + aoss_session_cost
total_session_cost_krw = total_session_cost_usd * KRW_RATE

print("=" * 60)
print("M4 OpenSearch Semantic Search — Cost Analysis")
print("=" * 60)
print(f"Instance type:       {INSTANCE_TYPE}")
print(f"Instance cost:       ${INSTANCE_COST_PER_HOUR:.2f}/hr")
print(f"Estimated time:      {estimated_total_min} min")
print(f"")
print(f"--- Compute ---")
print(f"  SageMaker:         ${compute_cost_usd:.3f} USD")
print(f"  OpenSearch (session): ${aoss_session_cost:.3f} USD")
print(f"  Session total:     ${total_session_cost_usd:.3f} USD ({total_session_cost_krw:.0f} KRW)")
print(f"")
print(f"--- OpenSearch Serverless Ongoing ---")
print(f"  Minimum OCUs:      {AOSS_MIN_INDEXING_OCU} indexing + {AOSS_MIN_SEARCH_OCU} search")
print(f"  Hourly minimum:    ${aoss_cost_per_hour:.2f}/hr")
print(f"  Monthly minimum:   ~${AOSS_MONTHLY_MIN:.0f}/month (while collection exists)")
print(f"  Monthly (KRW):     ~{AOSS_MONTHLY_MIN * KRW_RATE:,.0f} KRW/month")
print(f"")
print(f"--- vs NVIDIA Cosmos-embed NIM ---")
print(f"  NVAIE license:     order-of-thousands USD/yr per GPU node (list price varies)")
print(f"  OpenSearch (annual): ~${AOSS_MONTHLY_MIN * 12:,.0f}/yr")
print(f"  Savings:           Significant for workshop/dev use cases")
print(f"")
print(f"IMPORTANT: the collection bills ~24/7 until deleted. After the")
print(f"workshop, delete the collection AND its 3 policies to stop charges:")
print(f"  aws opensearchserverless delete-collection --id {collection['id']}")
print(f"  aws opensearchserverless delete-security-policy --name {COLLECTION_NAME}-enc --type encryption")
print(f"  aws opensearchserverless delete-security-policy --name {COLLECTION_NAME}-net --type network")
print(f"  aws opensearchserverless delete-access-policy   --name {COLLECTION_NAME}-access --type data")
print(f"  (or rely on user teardown / the workshop teardown script)")
print("=" * 60)
```
