## Pipeline Architecture

This lab implements the **8-stage** end-to-end Physical AI data pipeline from the
[AWS + NVIDIA AV 3.0 blog](https://aws.amazon.com/blogs/industries/building-an-end-to-end-physical-ai-data-pipeline-for-autonomous-vehicle-3-0-on-aws-with-nvidia/),
plus a few supplementary modules (M6, M11, M12) that extend it. Everything derives
from **three** prefixes in the shared S3 bucket: the **nuScenes-mini dataset**, the
flat **`model-cache/`** copies (read by M2 and M8 only) and the HuggingFace cache
**tree `hf-cache/`** (read by M5, M6 and M9). Arrows show data flow.

**Reading the diagram**
- **Three prefixes** in the shared bucket, not two: `datasets/nuscenes-mini/`,
  `model-cache/` and `hf-cache/`. Naming only the first two is how a new region gets
  seeded with `scripts/cache_models.sh` alone and comes up with M5/M6/M9/M10 broken —
  that happened in ap-northeast-2 on 2026-09-26, and `cache_models.sh` now warns about
  it in capitals. README step 6b seeds `hf-cache/`.
- **Core path:** M1 → M2 → M3 (ingest → caption → curate).
- **`m3/` has exactly one consumer: M12.** M5 and M6 read `m1/` plus the nuScenes
  images, and M9 reads `hf-cache/alpamayo-demo/` — none of the three touches `m3/`.
  An earlier version of this cell said M3 fanned out to all four, which sent anyone
  debugging M5 to look in `m3/` for an input that lives in `m1/`.
- **Model weights:** only **M2 and M8** `aws s3 sync` from `model-cache/`
  (`cosmos-reason1/`). M5, M6 and M9 restore the HuggingFace cache tree from
  `hf-cache/hub/` and point `HF_HOME` at it — `scripts/setup_cosmos_env.sh` does that.
  There is deliberately **no `model-cache/alpamayo-1.5/`**: a flat copy would be weight
  nothing reads.
- **M4** and **M3** both branch off **M2** (`m2/`); **M10** consumes **M9** (`m9/`)
  plus the admin's `m10-reference/`.
- **The search-driven loop is not wired here.** In the blog, Stage 5 *queries the
  Stage-4 index* to pick which clips to augment. In this lab M5/M6 read `m1/`
  directly and M4 is a leaf, so you can run either without the other.
- **M7** reads the **nuScenes CAM_FRONT images directly** (independent branch).
- **M8** reads `m1/manifest.json` plus the **nuScenes annotation tables** — its
  training targets are human labels, deliberately not M2's captions (training
  Cosmos Reason on its own output would make the loss self-referential).
- **M11** orchestrates the M1→M2→M3→M5 sub-pipeline as one SageMaker Pipeline.
